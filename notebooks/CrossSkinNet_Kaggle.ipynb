## 1. Code location
Implementation ZIP extract করে তার `CrossSkinNet` folder-এর path নিচে দিন। GitHub থেকে clone করলে আগে এই local commits push করতে হবে।

```python
from pathlib import Path
import subprocess, sys, os

REPO_DIR = Path.cwd()  # Set this to the extracted CrossSkinNet folder if needed.
assert (REPO_DIR / "scripts/00_smoke.py").is_file(), "Set REPO_DIR to the extracted implementation folder"
os.chdir(REPO_DIR)
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)
```

## 2. Tests
কোনো real dataset বা pretrained weight লাগবে না।

```python
subprocess.run([sys.executable, "-m", "pytest", "-q"], check=True)
```

## 3. Full synthetic flow
Merge থেকে evaluation, Grad-CAM ও SHAP পর্যন্ত চলবে। সব output config-এর output_root-এর নিচে থাকবে।

```python
subprocess.run([sys.executable, "scripts/00_smoke.py", "--config", "configs/default.yaml"], check=True)
```

## 4. Demo output দেখুন
নিচে generated metrics-এর path দেখাবে। এগুলো medical performance নয়।

```python
import yaml
config = yaml.safe_load(Path("configs/default.yaml").read_text())
output_root = Path(config["paths"]["output_root"])
for folder in sorted(output_root.glob("SYNTHETIC_DEMO_*")):
    print(folder / "outputs/results/tiny_finetuned/metrics.csv")
    print(folder / "outputs/results/tiny_finetuned/explanations")
```
